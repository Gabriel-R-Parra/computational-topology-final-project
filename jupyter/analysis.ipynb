# Analysis

```python
import polars as pl
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("darkgrid")
sns.set_context("notebook")
```

## Trials

A trial is defined as a tuple (subject, condition, activity).

There are 120 trials, 16 of which were collected from chronic stroke participants

In this analysis, we are interested in finding if the tsne representation of a trial's persistance diagram captures features that differentiate chronic stroke behaviors from healthy behaviors.

```python
tsne_0 = pl.scan_csv("model-results/tsne-0-trials.csv")
tsne_1 = pl.scan_csv("model-results/tsne-1-trials.csv")
```

### Activity Space

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(tsne_0.collect(), x="x", y="y", ax=axs[0], hue="activity", palette="tab10");
sns.scatterplot(tsne_1.collect(), x="x", y="y", ax=axs[1], hue="activity", palette="tab10");

axs[0].set_title("$H_0$")
axs[1].set_title("$H_1$")

fig.savefig("./figures/trials/activity.svg")
```

The 4th activity is a torque activity. The first 3 activites are forces activities.

It seems that the $H_0$ features were able to detect this difference.

### Condition Space

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(tsne_0.collect(), x="x", y="y", ax=axs[0], hue="condition", palette="tab10");
sns.scatterplot(tsne_1.collect(), x="x", y="y", ax=axs[1], hue="condition", palette="tab10");
```

### Healthy Space

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(tsne_0.collect(), x="x", y="y", ax=axs[0], hue="post-stroke", palette="tab10");
sns.scatterplot(tsne_1.collect(), x="x", y="y", ax=axs[1], hue="post-stroke", palette="tab10");
```

## Reps

A rep is defined as a tuple (subject, condition, activity, sub-activity, repetition).

There are 120 trials, 16 of which were collected from chronic stroke participants, in each of these trials, pariticipants were ordered to repeat two sub-activites 6 times each.

In this analysis, we are interested in finding if the tsne representation of a rep's persistance diagram captures features that differentiate chronic stroke behaviors from healthy behaviors.

```python
tsne_0 = pl.scan_csv("./model-results/tsne-0-reps.csv")
tsne_1 = pl.scan_csv("./model-results/tsne-1-reps.csv")
```

### Activity Space

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(tsne_0.collect(), x="x", y="y", ax=axs[0], hue="activity", palette="tab10");
sns.scatterplot(tsne_1.collect(), x="x", y="y", ax=axs[1], hue="activity", palette="tab10");
```

### Condition Space

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(tsne_0.collect(), x="x", y="y", ax=axs[0], hue="condition", palette="tab10");
sns.scatterplot(tsne_1.collect(), x="x", y="y", ax=axs[1], hue="condition", palette="tab10");
```

### Healthy Space

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(tsne_0.collect(), x="x", y="y", ax=axs[0], hue="post-stroke", palette="tab10");
sns.scatterplot(tsne_1.collect(), x="x", y="y", ax=axs[1], hue="post-stroke", palette="tab10");

axs[0].set_title("$H_0$")
axs[1].set_title("$H_1$")
fig.savefig("./figures/reps/post-stroke.svg")
```

### Participant Space

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(tsne_0.collect(), x="x", y="y", ax=axs[0], hue="participant", palette="tab20");
sns.scatterplot(tsne_1.collect(), x="x", y="y", ax=axs[1], hue="participant", palette="tab20");
```

### Task Space

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(tsne_0.filter().collect(), x="x", y="y", ax=axs[0], hue="sub_task", palette="tab10");
sns.scatterplot(tsne_1.filter().collect(), x="x", y="y", ax=axs[1], hue="sub_task", palette="tab10");
```
